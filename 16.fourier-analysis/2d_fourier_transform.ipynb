The **Discrete Fourier Transform** (DFT) transforms an image
$\ell\left[n,m \right]$, of finite size $N \times M$, into the complex
image Fourier transform $\mathscr{L}\left[u,v \right]$ as:
$$\mathscr{L}\left[u,v \right] =  
\mathcal{F} \left\{ \ell\left[n,m \right] \right\}
=
\sum_{n=0}^{N-1} \sum_{m=0}^{M-1} \, \ell\left[n,m \right] 
\exp{ \left( -2\pi j \left( \frac{u\, n}{N} + \frac{v\, m}{M} \right) \right)}
$${#eq-fourier} We will call $\mathscr{L}\left[u,v \right]$ the Fourier
transform of $\ell\left[m,n \right]$. We will often represent the
relationship between the signal as its transform as:
$$\ell\left[n,m \right] \xrightarrow{\mathscr{F}} \mathscr{L}\left[u,v \right]$$

```python
# lets redefine everything we had to match the equation

# first, for simplicity, we will import the image as monochromatic. 
# we will also flip the rows, since we define 0,0 as bottom left, 
# while the original image defines 0,0 as top left.
# we shall also significantly reduce the image size to accomodate for the (for now)
# inefficient fourier transform pipeline
l = cv2.resize((cv2.imread("nala_ares.jpg", cv2.IMREAD_GRAYSCALE)[::-1, :]), (320, 210))
px.imshow(l, color_continuous_scale = 'gray', origin = 'lower')
```

```python
# lets now also define N and M
M, N = l.shape
pp(N)
pp(M)
```

```python
def get_L(u, v):
  outer_sum = 0
  for n in np.arange(int(-N/2), int(N/2), 1):
    inner_sum = 0
    for m in np.arange(int(-M/2), int(M/2), 1):
      # we call l by m then n because the image is ordered in rows first then columns
      F = l[m][n] * np.exp((-2j * np.pi) * (((u * n) / N) + ((v * m) / M)))
      inner_sum += F
    outer_sum += inner_sum
  return outer_sum
```

```python
L = None
for u in np.arange(-8.0, 8.0, 0.2):
  col_L = None
  for v in np.arange(-8.0, 8.0, 0.2):
    pp(f"{u}, {v}")
    _L = get_L(u, v)
    col_L = _L if col_L is None else np.append(col_L, _L)
  L = [col_L] if L is None else np.vstack([[col_L], L])
  
```
