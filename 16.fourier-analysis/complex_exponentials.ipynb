```python
import numpy as np
import plotly.express as px
```

## exploring complex exponentials

```python
# lets describe t as a number line for seconds since wave started (assuming dimension as temporal)
t = np.linspace(0, 4, 256)

# lets now define frequency as the amount of wave per second (Hz)
f = 1

# we use np.exp() instead of math.exp() because math.exp() cannot take in imaginary exponents,
# this also makes everything neater (everything revolves around numpy)
wave = np.exp(t * 2 * np.pi * f * 1j)
# declaring wave as:
#   wave = [np.exp(_t * 1j) for _t in t]
# might seem as the intuitive option. however, this results in the output being a list, 
# which would then need to be converted back to np array 
# 
# theres also some computational inefficiencies with using this format, although i am not yet
# able to confidently assess it in this comment yet.
# 
# in good practice though, sticking to a more brief call in numpy is generally more efficient,
# both in terms of readibility and computation

print(type(wave))
# print(wave)
```

```python
# lets now plot the `wave` variable in 3d

# we use px.line3d to define a graph in 3-dimensional axis
graph = px.line_3d(
  x = wave.real,
  z = wave.imag,
  y = t
)

# lets change the perspective type to ortographic to better interact with the visual properties
# of the graph
graph.update_scenes(camera_projection_type = "orthographic")

graph.show()
```

```python
# now, let us create a sine function of arbitrary period
sinewave_f = 2
sinewave = np.sin(2 * np.pi * sinewave_f * t)

graph = px.line(x = t, y = sinewave)
graph.show()
```

```python
# we now have 2 things: a spiral in 3d, and a wave in 2d. lets now scale the value of each instance
# of the 3d spiral against the 2d wave

compound_wave = sinewave * wave

graph = px.line_3d(
  x = compound_wave.real,
  z = compound_wave.imag,
  y = t
)

graph.update_scenes(camera_projection_type = "orthographic")

graph.show()
```
