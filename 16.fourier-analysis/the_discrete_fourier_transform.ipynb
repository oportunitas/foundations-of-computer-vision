```python
import numpy as np
import torch as pt
import plotly.express as px
from pprint import pp
```

lets first try to create a 1-dimensional discrete fourier transform function

```python
# since this is discrete, f is an array of f(x), not an actual function
def discrete_fourier_transform(f, x):
  n = len(x)
  res = None
  # u is the frequency of the decomposed wave
  for u in np.linspace(-32.0, 32.0, 65):
    sum = 0.0
    for _f, _x in zip(f, x):
      sum += _f * np.exp((-2j * np.pi) * ((u * _x) / n))
    res = [u, sum] if res is None else np.vstack([[u, sum], res])
  return res
```

```python
x = np.linspace(-32.0, 32.0, 65)
print(x)
ramp_signal = lambda x: 0.5 * x

graph = px.line(x = x, y = ramp_signal(x))
graph.show()
```

```python
ramp_signal_fourier = discrete_fourier_transform(ramp_signal(x), x)
graph = px.line_3d(
  x = ramp_signal_fourier[:, 0].real, 
  z = ramp_signal_fourier[:, 1].real, 
  y = ramp_signal_fourier[:, 1].imag)
graph.update_scenes(camera_projection_type = "orthographic")
graph.show()
```

```python
# again, since this is discrate, f and x are arrays of equal length
def discrete_inverse_fourier_transform(f, x):
  n = len(x)
  res = None
  for u in np.linspace(-32.0, 32.0, 65):
    sum = 0.0
    for _f, _x in zip(f, x):
      sum += _f * np.exp((+2j * np.pi) * ((u * _x) / n))
    _res = [u, (sum / n)]
    # pp(_res)
    # pp(res)
    # pp("")
    res = _res if res is None else np.vstack([_res, res])
  return res
```

```python
# pp(ramp_signal_fourier[:, 1])
ramp_signal_fourier_inverse = discrete_inverse_fourier_transform(ramp_signal_fourier[:, 1], ramp_signal_fourier[:, 0])
# pp(ramp_signal_fourier_inverse)
graph = px.line_3d(
  x = ramp_signal_fourier_inverse[:, 0].real, 
  y = ramp_signal_fourier_inverse[:, 1].real, 
  z = ramp_signal_fourier_inverse[:, 1].imag)
graph.update_scenes(camera_projection_type = "orthographic")
graph.show()
```
