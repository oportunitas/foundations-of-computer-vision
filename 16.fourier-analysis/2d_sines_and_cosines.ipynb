## 1d waves

```python
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from pprint import pp
```

lets create a discrete sine wave function, according to this definition:
$$s_k\left[n\right] = \sin\left( \frac{2 \pi}{N} \, k \, n \right)$$

```python
# N ∈ ℕ, n ∈ [0, N - 1], k ∈ [1, N/2]. where:
#   N is the array size
#   n is an iteration of the array,
#   k is the frequency of the wave over N
def make_disc_sine(N, k):
  def res(n):
    return np.sin(((2 * np.pi) / N) * k * n)
  return res

N = 20
k = 2
# this means that this wave (very layman's term) goes N units to complete k time
disc_sine = make_disc_sine(N, k)
x = np.linspace(0, N, N + 1)

graph = px.scatter(x = x, y = disc_sine(x))
graph.show()
```

## 2d waves

> The same analysis can be extended to two dimensions (2D). In 2D, the
> discrete sine and cosine waves are as follows:
> 
> $$s_{u,v}\left[n,m\right] = A \sin \left(2 \pi \left( \frac{u\,n}{N}  + \frac{v\,m}{M}  \right) \right)$$
> $$c_{u,v}\left[n,m\right] = A \cos \left(2 \pi \left( \frac{u\,n}{N}  + \frac{v\,m}{M}  \right) \right)$$
> 
> where $A$ is the amplitude and $u$ and $v$ are the two spatial
> frequencies and define how fast or slow the waves change along the
> spatial dimensions $n$ and $m$.

lets try to implement the sin() function

```python
# if we were to describe n and m as orthogonal on 2d plane:
#   A: wave amplitude
#   N: sample size of wave in n axis
#   u: wave frequency in n per sample size in n
#   M: sample size of wave in m axis
#   v: wave frequency in m per sample size in m
def make_disc_sine_2d(A, N, u, M, v):
  def disc_sine_2d(n, m):
    return A * np.sin((np.pi * 2) * (((u * n) / N) + ((v * m) / M)))
  return disc_sine_2d

A = 1
N = 20
M = 20
n = np.linspace(0, N, 4 * N + 1)
u = 7
m = np.linspace(0, M, 4 * M + 1)
v = -5

disc_sine_2d = make_disc_sine_2d(A, N, u, M, v)

graph = go.Figure(go.Surface(x = n, y = m, z = [[disc_sine_2d(_n, _m) for _n in n] for _m in m]))
graph.update_scenes(camera_projection_type = "orthographic")
graph.update_layout(autosize = True)
graph.show()
```
