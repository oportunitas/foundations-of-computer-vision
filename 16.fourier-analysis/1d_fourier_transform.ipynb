```python
import plotly.express as px
import numpy as np
from pprint import pp
from scipy.fft import fft
```

```python
def square_wave(x: int) -> int:
  return np.sin(x) - 0.5 * np.sin(x * 12)
  return (0 if int(x) % 2 else 1)
vectorized_square_wave = np.vectorize(square_wave)
```

```python
square_wave_x = np.linspace(0, 10, 10000)
square_wave_y = vectorized_square_wave(square_wave_x)
```

```python
px.line(x = square_wave_x, y = square_wave_y)
```

```python
def fourier_transform(x_arr: np.ndarray, y_arr: np.ndarray, w: np.float64) -> complex:
  return np.sum(y_arr * np.exp(-2j * np.pi * w * (x_arr / len(x_arr))))
```

```python
def full_fourier_transform(y_arr: np.ndarray) -> np.ndarray:
  N = len(y_arr)
  n = np.arange(N)
  k = n.reshape((N, 1)) # column vector
  
  e = np.exp(-2j * np.pi * k * (n / N))
  return np.dot(e, y_arr)
```

```python
ft_square_wave_y = full_fourier_transform(square_wave_y)
```

```python
fig = px.scatter_3d(z = square_wave_x, x = ft_square_wave_y.real, y = ft_square_wave_y.imag)
fig.update_layout(scene_camera_projection_type='orthographic')
fig.show()
```

```python
def full_inverse_fourier_transform(y_arr: np.ndarray) -> np.ndarray:
  N = len(y_arr)
  n = np.arange(N)
  k = n.reshape((N, 1))

  e = np.exp(2j * np.pi * k * (n / N))
  return np.dot(e, y_arr) / N
```

```python
inv_ft_square_wave_y = full_inverse_fourier_transform(ft_square_wave_y)
```

```python
px.line(x = square_wave_x, y = inv_ft_square_wave_y.real)
```
